```python
def get_center_range(ne, sw):
    ne, sw = ne.split(", "), sw.split(", ")
    ne[0], ne[1] = float(ne[1][:-2]), float(ne[0][:-2])
    sw[0], sw[1] = float(sw[1][:-2]), float(sw[0][:-2])
    cx, cy = (np.array(ne) + np.array(sw)) / 2
    rx, ry = ne[0] - cx, ne[1] - cy
    return (cx, cy), (rx, ry)
```

```python
import numpy as np

def mse(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return np.mean((y_true - y_pred) ** 2)

def rmse(y_true, y_pred):
    return np.sqrt(mse(y_true, y_pred))

def mae(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    return np.mean(np.abs(y_true - y_pred))

def mape(y_true, y_pred):
    y_true = np.array(y_true) + 1
    y_pred = np.array(y_pred) + 1
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

def smape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    denominator = np.abs(y_true) + np.abs(y_pred)
    diff = np.abs(y_true - y_pred)
    smape_values = np.where(denominator != 0, 2 * diff / denominator, 0)
    return np.mean(smape_values) * 100

def accuracy_from_mape(mape_val):
    """
    将 MAPE 直接转换为准确率
    """
    return 100 - mape_val

def accuracy_from_mae(mae_val, y_true):
    """
    基于 MAE 和真实值的均值计算百分比准确率
    """
    y_true = np.array(y_true)
    mean_true = np.mean(np.abs(y_true))
    return (1 - mae_val / mean_true) * 100

def accuracy_from_rmse(rmse_val, y_true):
    """
    基于 RMSE 和真实值的均值计算百分比准确率
    """
    y_true = np.array(y_true)
    mean_true = np.mean(np.abs(y_true))
    return (1 - rmse_val / mean_true) * 100
```

```python
def get_acc(y_true, y_pred):
    mse_val = mse(y_true, y_pred)
    rmse_val = rmse(y_true, y_pred)
    mae_val = mae(y_true, y_pred)
    mape_val = mape(y_true, y_pred)
    smape_val = smape(y_true, y_pred)

    print("MSE:", mse_val)
    print("RMSE:", rmse_val)
    print("MAE:", mae_val)
    print("MAPE:", mape_val, "%")
    print("SMAPE:", smape_val, "%")

    # 转换为百分比准确率
    print("基于 MAPE 的准确率:", accuracy_from_mape(mape_val), "%")
    print("基于 MAE 的准确率:", accuracy_from_mae(mae_val, y_true), "%")
    print("基于 RMSE 的准确率:", accuracy_from_rmse(rmse_val, y_true), "%")
```
